**3. header setup**

```python
# header ini wajib ada biar server tidak menganggap kita bot/scraper
HEADERS = {
    # identitas browser yang kita pura-purakan
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/124.0.0.0 Safari/537.36"
    ),
    # bahasa yang kita minta dari server
    "Accept-Language": "en-US,en;q=0.9",
}

# url dasar yang akan kita scrapping
BASE_URL = "https://www.freelancer.com/jobs"

print("Setup selesai")
```

**Penjelasan :**
>Kode ini digunakan untuk melakukan konfigurasi awal web scraping. Variabel `HEADERS` berisi identitas browser (User-Agent) dan bahasa yang digunakan agar server menganggap permintaan berasal dari pengguna biasa, bukan bot atau scraper. Variabel `BASE_URL` menyimpan alamat website yang akan diambil datanya, yaitu halaman lowongan kerja [freelancer](https://Freelancer.com). Perintah `print("Setup selesai")` digunakan untuk menampilkan pesan bahwa proses konfigurasi telah berhasil dilakukan dan program siap dijalankan.

**4. tes koneksi**

```python
# kirim get requests ke url dasar untuk memastikan koneksi
# timeout=10 -> waktu tunggu jika koneksi tidak berhasil
res = requests.get(BASE_URL, headers=HEADERS, timeout=10)

# cek apakah koneksi berhasil (200 = ok, 403 = diblokir, 429 = rate limit)
print(f"Status: {res.status_code}")
```

**Penjelasan :**
> Kode ini digunakan untuk mengirim permintaan (GET request) ke website yang telah ditentukan pada `BASE_URL` menggunakan `header` yang telah dibuat sebelumnya. Parameter `timeout=10` memberikan batas waktu tunggu selama 10 detik jika koneksi gagal. Setelah itu, `res.status_code` digunakan untuk menampilkan kode status respons dari server guna memastikan apakah koneksi berhasil, misalnya kode 200 berarti berhasil, 403 berarti akses ditolak, dan 429 berarti terlalu banyak permintaan (rate limit).

**4. melihat html mentah**

```python
# parse html dari response menggunakan html.panser bawaan python
soup = BeautifulSoup(res.text, "html.parser")

# Lihat 3000 karakter pertama buat orientasi struktur
print(soup.prettify()[:3000])
```

**Penjelasan :**
> Kode ini digunakan untuk mengubah isi HTML yang diperoleh dari website menjadi objek `BeautifulSoup` agar lebih mudah dibaca dan diproses. Parser `html.parser` digunakan untuk menerjemahkan struktur HTML. Selanjutnya, `soup.prettify()` digunakan untuk menampilkan kode HTML yang sudah dirapikan, dan `[:3000]` membatasi output hanya pada 3000 karakter pertama agar memudahkan proses pengamatan struktur halaman sebelum melakukan scraping data yang dibutuhkan.

**Insight :**
> Tampilan sebagian struktur HTML digunakan untuk memastikan bahwa data yang diterima sesuai dengan halaman website target dan membantu proses identifikasi tag, class, serta selector yang akan digunakan dalam scraping.

**5. melihat selector yang tepat**

```python
# Coba beberapa selector umum Freelancer
candidates = [
    "div.JobSearchCard-item",
    "div[class*='JobSearchCard']",
]

for selector in candidates:
    found = soup.select(selector)
    print(f"'{selector}' → {len(found)} elemen ditemukan")
```

**Penjelasan :**
> Kode ini digunakan untuk mencoba beberapa selector HTML yang kemungkinan berisi data lowongan kerja pada website Freelancer. Setiap selector disimpan dalam list `candidates`, kemudian dilakukan pencarian menggunakan `soup.select()`. Hasilnya ditampilkan dalam bentuk jumlah elemen yang ditemukan untuk membantu menentukan selector yang paling tepat digunakan dalam proses scraping data.

**Insight :**
> Hasil pengujian menunjukkan bahwa selector `div.JobSearchCard-item` menghasilkan 50 elemen yang sesuai dengan jumlah proyek yang ditampilkan pada halaman. Sementara itu, selector `div[class*="JobSearchCard"]` menghasilkan 631 elemen karena mencakup seluruh komponen yang berkaitan dengan kartu proyek. Oleh sebab itu, selector `div.JobSearchCard-item` dipilih karena lebih spesifik dan efektif untuk proses pengambilan data proyek.

**6. function scrape satu halaman**

```python
def scrape_page(page_num: int) -> list[dict]:
    """
    Scrape satu halaman job listing dari Freelancer.com.

    Parameter:
        page_num (int): nomor halaman (1, 2, 3, ...)

    Return:
        list of dict: daftar job yang berhasil di-scrape
    """

    # Tentukan URL: halaman 1 pakai BASE_URL, halaman 2+ tambahkan nomor halaman
    url = BASE_URL if page_num == 1 else f"{BASE_URL}/{page_num}"

    # Kirim request dengan error handling
    try:
        res = requests.get(url, headers=HEADERS, timeout=10)
        # Kalau status bukan 2xx (misal 403, 429), langsung raise error
        res.raise_for_status()
    except requests.RequestException as e:
        # Tangkap semua error koneksi dan cetak pesannya
        print(f"Gagal fetch halaman {page_num}: {e}")
        return []  # return list kosong agar program tetap jalan

    # Parse HTML hasil response
    soup = BeautifulSoup(res.text, "html.parser")

    # Ambil semua elemen job card berdasarkan CSS selector
    # Ganti selector ini sesuai hasil Cell 4 kalau perlu
    job_cards = soup.select("div.JobSearchCard-item")

    # Kalau tidak ada job card ditemukan, beri peringatan
    if not job_cards:
        print(f"Halaman {page_num}: tidak ada job card ditemukan")
        return []

    jobs = []  # list kosong untuk tampung hasil tiap halaman

    for card in job_cards:

        # ── Judul & URL ──────────────────────────────────────────
        title_tag = card.select_one("a.JobSearchCard-primary-heading-link")
        # get_text(strip=True) → ambil teks bersih tanpa spasi/newline di ujung
        title = title_tag.get_text(strip=True) if title_tag else None
        # Tambahkan domain karena href-nya relatif (contoh: /projects/xxx)
        link  = "https://www.freelancer.com" + title_tag["href"] if title_tag else None

        # ── Deskripsi ─────────────────────────────────────────────
        desc_tag    = card.select_one("p.JobSearchCard-primary-description")
        description = desc_tag.get_text(strip=True) if desc_tag else None

        # ── Skills/Tags ───────────────────────────────────────────
        # select() → ambil SEMUA elemen yang cocok (bisa banyak)
        skill_tags = card.select("a.JobSearchCard-primary-tagsLink")
        # List comprehension: ambil teks dari tiap tag skill
        skills     = [s.get_text(strip=True) for s in skill_tags]

        # ── Budget / Average Bid ──────────────────────────────────
        budget_tag = card.select_one("div.JobSearchCard-primary-price")
        budget     = budget_tag.get_text(strip=True) if budget_tag else None

        # ── Jumlah Bids ───────────────────────────────────────────
        bids_tag = card.select_one("div.JobSearchCard-secondary-entry")
        bids     = bids_tag.get_text(strip=True) if bids_tag else None

        # ── Sisa Waktu ────────────────────────────────────────────
        time_tag  = card.select_one("span.JobSearchCard-primary-heading-days")
        time_left = time_tag.get_text(strip=True) if time_tag else None

        # Masukkan semua data satu job ke dalam dict, lalu append ke list
        jobs.append({
            "title":       title,
            "link":        link,
            "description": description,
            "skills":      ", ".join(skills),  # gabung list jadi string dipisah koma
            "budget":      budget,
            "bids":        bids,
            "time_left":   time_left,
        })

    print(f"Halaman {page_num}: {len(jobs)} job ditemukan")
    return jobs


# ── Test dulu 1 halaman sebelum scrape banyak ─────────────────
sample = scrape_page(1)
print(f"\nContoh data pertama:")
print(sample[0] if sample else "Kosong — cek selector!")
```

**Penjelasan :**
> Kode ini membuat function `scrape_page()` yang bertugas mengambil data lowongan kerja dari satu halaman Freelancer.com. Fungsi akan mengirim request ke website, memeriksa apakah koneksi berhasil, kemudian mengubah HTML menjadi objek `BeautifulSoup` untuk diproses. Selanjutnya, program mengambil informasi penting dari setiap lowongan seperti judul, tautan, deskripsi, skill yang dibutuhkan, budget, jumlah penawaran (bids), dan sisa waktu. Data tersebut disimpan dalam bentuk dictionary dan dikumpulkan ke dalam list. Di bagian akhir, fungsi diuji dengan mengambil data dari halaman pertama dan menampilkan satu contoh hasil scraping untuk memastikan proses berjalan dengan benar.

**7. scrape multi halaman**

```python
def scrape_all(max_pages: int = 5, delay: float = 2.0) -> pd.DataFrame:
    """
    Scrape beberapa halaman sekaligus dan gabung hasilnya ke DataFrame.

    Parameter:
        max_pages (int): jumlah halaman yang mau di-scrape
        delay (float): jeda antar request dalam detik (default 2 detik)

    Return:
        pd.DataFrame: semua job dari semua halaman
    """
    all_jobs = []  # tampung semua job dari semua halaman

    for page in range(1, max_pages + 1):
        jobs = scrape_page(page)    # scrape 1 halaman
        all_jobs.extend(jobs)       # tambahkan hasilnya ke list utama

        # Jeda setelah setiap halaman KECUALI halaman terakhir
        # Ini penting agar tidak kena rate-limit / IP ban dari server
        if page < max_pages:
            time.sleep(delay)

    # Konversi list of dict menjadi DataFrame pandas
    df = pd.DataFrame(all_jobs)
    print(f"\n Total: {len(df)} job dari {max_pages} halaman")
    return df


# Mulai scrape 10 halaman dulu (500 job), naikkan kalau sudah yakin berjalan
df = scrape_all(max_pages=10)

# Tampilkan preview DataFrame di Colab
df.head()
```

**Penjelasan :**
> Kode ini membuat functiom `scrape_all()` untuk mengambil data lowongan kerja dari beberapa halaman sekaligus. Function akan memanggil `scrape_page()` pada setiap halaman, menggabungkan seluruh hasil ke dalam satu list, dan memberikan jeda antar request menggunakan `time.sleep()` agar tidak dianggap spam oleh server. Setelah semua data terkumpul, data dikonversi menjadi DataFrame menggunakan pandas sehingga lebih mudah dianalisis. Pada bagian akhir, fungsi dijalankan untuk mengambil data dari 10 halaman dan menampilkan beberapa baris pertama hasil scraping sebagai pratinjau data.

**4. Data Transformation**

```python
# mengelompokkan data yang akan di digunakan
import re
# LANGKAH 1: DATA CLEANING (Ekstraksi Anggaran)
def extract_budget(text):
    if pd.isna(text):
        return None

    # Menemukan semua angka di teks anggaran
    numbers = re.findall(r'\d+', str(text))

    if len(numbers) == 0:
        return None

    # Mengubah teks angka menjadi integer
    numbers = [int(x) for x in numbers]

    # Mengambil rata-rata jika bentuknya range (misal: 3.000.000 - 5.000.000)
    return sum(numbers) / len(numbers)

# Terapkan pembersihan anggaran (Ganti 'anggaran' dengan nama kolom anggaran aslimu)
# Di sini kita asumsikan namanya 'anggaran' atau 'budget'
nama_kolom_anggaran = 'anggaran' if 'anggaran' in df.columns else 'budget_clean'
df['budget_clean'] = df[nama_kolom_anggaran].apply(extract_budget)

# Drop data yang tidak memiliki anggaran atau keahlian untuk akurasi analisis
df_cleaned = df.dropna(subset=['keahlian', 'budget_clean']).copy()

# LANGKAH 2: DATA TRANSFORMATION (Memisahkan Skill)
# Mengubah string berisi koma menjadi sebuah List
df_cleaned['skill_list'] = df_cleaned['keahlian'].str.split(', ')

# Melakukan 'explode' agar setiap skill memiliki barisnya masing-masing
df_exploded = df_cleaned.explode('skill_list')

# Membersihkan spasi di awal/akhir teks skill yang tersisa
df_exploded['skill_list'] = df_exploded['skill_list'].str.strip()

# LANGKAH 3: ANALISIS & AGREGASI
# Mengelompokkan berdasarkan skill untuk menghitung: Total Proyek & Rata-rata Anggaran
skill_analysis = df_exploded.groupby('skill_list').agg(
    jumlah_proyek=('budget_clean', 'count'),
    rata_rata_anggaran=('budget_clean', 'mean')
).reset_index()

# Urutkan berdasarkan jumlah proyek terbanyak (paling dicari)
skill_analysis = skill_analysis.sort_values(by='jumlah_proyek', ascending=False)

# LANGKAH 4: MENAMPILKAN HASIL
print("=== SKILL PALING DICARI & BERNILAI TINGGI BERDASARKAN JUMLAH PROYEK & ANGGARAN ===")
print(skill_analysis.head(25).to_string(index=False))
```

Penjelasan:

kode ini berfungsi untuk mencari tahu keahlian (skill) apa saja yang paling populer (paling banyak dicari) dan memiliki nilai pasar tertinggi (rata-rata anggaran terbesar) berdasarkan dataset diatas.

Langkah 1: Pembersihan Data (Data Cleaning & Extraction)
Data mentah dari internet atau database sering kali kotor. Kolom anggaran biasanya ditulis dalam bentuk teks manual (misal: "Rp 5.000.000 - Rp 7.000.000" atau "Budget 3jt"). Komputer tidak bisa menghitung rata-rata dari teks seperti itu.

1. Membuat Mesin Pengekstrak (extract_budget): Kode ini membuat fungsi kustom
menggunakan Regular Expression (Regex). Regex mendeteksi karakter digital (\d+) di dalam teks.

2. Menangani Rentang Nilai (Range): Jika teks bertuliskan rentang seperti 3000000 - 5000000, Regex akan menangkap dua angka: [3000000, 5000000]. Kode kemudian menjumlahkannya dan membaginya dengan total item (sum / len), sehingga menghasilkan nilai tengah/rata-rata, yaitu 4000000.

3. Standarisasi Kolom: Kode secara pintar mengecek apakah kolom di dataset Anda bernama 'anggaran' atau bukan. Setelah itu, fungsi pembersih tadi diterapkan ke seluruh baris.

4. Eliminasi Data Kosong (.dropna): Untuk menjaga akurasi analisis, baris yang tidak memiliki informasi keahlian atau tidak memiliki angka anggaran langsung dibuang.

Langkah 2: Transformasi Data (Data Transformation)
Masalah utama dalam analisis data teks adalah data multiline dalam satu sel. Kolom 'keahlian' Anda kemungkinan besar berisi banyak skill yang dipisahkan koma, contohnya: Python, Data Analysis, SQL. Jika langsung dikelompokkan, Python akan menganggap "Python, Data Analysis, SQL" sebagai satu kesatuan skill yang unik.

1. Konversi ke List (.str.split): Teks "Python, Data Analysis, SQL" diubah menjadi struktur data list: ['Python', 'Data Analysis', 'SQL'].

2. Metode Explode (.explode): Ini adalah bagian paling krusial. Fungsi explode() akan membongkar list tersebut. Jika Proyek ID 001 membutuhkan 3 skill di atas, maka setelah di-explode, data akan berubah menjadi 3 baris:

Langkah 3: Analisis & Agregasi (Aggregation)
Setelah data rapi di mana satu baris hanya mewakili satu keahlian dan satu proyek, barulah analisis statistik dilakukan.

1. Pengelompokkan (.groupby): Data dikumpulkan berdasarkan nama keahlian yang sama.

2. Kombinasi Statistik (.agg): Kode menjalankan dua fungsi kalkulasi sekaligus pada kolom anggaran (budget_clean):

>count: Menghitung berapa kali skill tersebut muncul (merepresentasikan Jumlah Proyek).

>mean: Menghitung rata-rata nilai anggaran dari semua proyek yang membutuhkan skill tersebut (merepresentasikan Nilai Pasar/Harga Skill).

3. Pengurutan (.sort_values): Tabel diurutkan dari skill yang paling banyak muncul di baris teratas untuk mengetahui keahlian apa yang paling populer di pasar saat ini.

**4. visualisasi skill dengan anggaran tertinggi**

```python
# skill dengan rata-rata anggaran tertinggi
# Filter dulu: minimal 5 proyek biar datanya representatif
top_value = (
    df_visualisasi[df_visualisasi["jumlah_proyek"] >= 5]
    .nlargest(5, "rata_rata_anggaran")
    .sort_values("rata_rata_anggaran")
)

fig_value = px.bar(
    top_value,
    x="rata_rata_anggaran",
    y="skill_list",
    orientation="h",
    title="Skill IT dengan Rata-Rata Anggaran Tertinggi",
    labels={
        "rata_rata_anggaran": "Rata-Rata Anggaran (USD)",
        "skill_list":         "Skill"
    },
    color="rata_rata_anggaran",
    color_continuous_scale="Greens",
    text=top_value["rata_rata_anggaran"].round(0).astype(int),
)

fig_value.update_traces(texttemplate="$%{text}", textposition="outside")
fig_value.update_layout(
    height=600,
    coloraxis_showscale=False,
    plot_bgcolor="#0f0f0f",
    paper_bgcolor="#1a1a1a",
    font_color="white",
    title_font_size=16,
)

fig_value.show()
```

Visualisasi ini dibuat menggunakan Plotly Express dengan tema gelap (*dark mode*), di mana setiap batang memiliki gradasi warna hijau berdasarkan nilai anggarannya, serta dilengkapi dengan label teks nominal dolar (di-bulatkan tanpa desimal) yang diposisikan di luar batang grafik agar mempermudah pembacaan data secara instan.

**5. visualisasi peta skill**

```python
df_filtered = df_visualisasi[df_visualisasi["jumlah_proyek"] >= 5].copy()

median_demand = df_filtered["jumlah_proyek"].median()
median_budget = df_filtered["rata_rata_anggaran"].median()

# Pilih skill yang layak diberi label:
# top 10 demand + top 10 value, gabung dan hapus duplikat
top_label = pd.concat([
    df_filtered.nlargest(5, "jumlah_proyek"),
    df_filtered.nlargest(5, "rata_rata_anggaran"),
]).drop_duplicates(subset="skill_list")

# Kolom label: hanya muncul untuk skill terpilih, sisanya kosong
df_filtered["label"] = df_filtered["skill_list"].where(
    df_filtered["skill_list"].isin(top_label["skill_list"]), ""
)

fig_scatter = px.scatter(
    df_filtered,
    x="jumlah_proyek",
    y="rata_rata_anggaran",
    text="label",                              # hanya top skill yang punya label
    title="Peta Skill IT: Volume Proyek vs Rata-Rata Anggaran",
    labels={
        "jumlah_proyek":      "Jumlah Proyek",
        "rata_rata_anggaran": "Rata-Rata Anggaran (USD)"
    },
    size="jumlah_proyek",
    size_max=40,
    color="rata_rata_anggaran",
    color_continuous_scale="Viridis",
    hover_name="skill_list",                   # semua skill tetap keliatan saat hover
    hover_data={
        "jumlah_proyek":      True,
        "rata_rata_anggaran": ":.2f",
        "skill_list":         False,
        "label":              False,           # sembunyiin kolom label dari hover
    }
)

# Garis median pembatas kuadran
fig_scatter.add_vline(x=median_demand, line_dash="dash", line_color="gray", opacity=0.5)
fig_scatter.add_hline(y=median_budget, line_dash="dash", line_color="gray", opacity=0.5)

# Label kuadran
fig_scatter.add_annotation(text="⭐ Sweet Spot",    x=df_filtered["jumlah_proyek"].max()*0.75, y=df_filtered["rata_rata_anggaran"].max()*0.95, showarrow=False, font=dict(color="gold",       size=12))
fig_scatter.add_annotation(text="💎 Niche Mahal",   x=df_filtered["jumlah_proyek"].max()*0.05, y=df_filtered["rata_rata_anggaran"].max()*0.95, showarrow=False, font=dict(color="lightblue",  size=12))
fig_scatter.add_annotation(text="📈 Volume Tinggi", x=df_filtered["jumlah_proyek"].max()*0.75, y=df_filtered["rata_rata_anggaran"].max()*0.10, showarrow=False, font=dict(color="lightgreen", size=12))
fig_scatter.add_annotation(text="📉 Low Priority",  x=df_filtered["jumlah_proyek"].max()*0.05, y=df_filtered["rata_rata_anggaran"].max()*0.10, showarrow=False, font=dict(color="salmon",     size=12))

fig_scatter.update_traces(textposition="top center", textfont_size=9)
fig_scatter.update_layout(
    height=650,
    plot_bgcolor="#0f0f0f",
    paper_bgcolor="#1a1a1a",
    font_color="white",
    title_font_size=16,
)

fig_scatter.show()
```

Kode tersebut berfungsi untuk menyaring data *skill* (minimal 5 proyek) dan menghitung nilai tengah (median) sebagai batas wilayah, lalu memetakan seluruh *skill* tersebut ke dalam grafik *scatter plot* interaktif berbasis tema gelap (*dark mode*). Grafik ini membagi *skill* IT ke dalam empat kuadran analisis strategis berdasarkan volume proyek dan anggaran (yaitu *Sweet Spot*, *Niche Mahal*, *Volume Tinggi*, dan *Low Priority*), di mana ukuran serta warna lingkaran otomatis membesar mengikuti data, sementara label teks nama *skill* sengaja dibatasi hanya muncul pada 5 *skill* terpopuler dan 5 *skill* termahal agar tampilan visual tetap rapi, bersih, dan mudah dipahami.
