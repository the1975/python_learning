## Data Cleaning
1 . Cari Missing Value masing-masing dataset

Tentukan strategi untuk tiap kolom pada masing-masing dataset:

- Hapus?
- Isi?
- Biarkan?

````python
# Strategi per kolom
# biarkan
df['director'].fillna('Unknown', inplace=True)
df['cast'].fillna('Unknown', inplace=True)
df['country'].fillna('Unknown', inplace=True)
df['duration'].fillna('Unknown', inplace=True)

# hapus
df.dropna(subset=['date_added'], inplace=True)
df = df.dropna(subset=['date_added'])

# isi
df['rating'].fillna('df[rating].mode()[0]', inplace=True)
````

````python
# Cek hasilnya
df['date_added'].head()

print(df.isnull().sum())
````

Jawaban disini

Strategi penanganan missing value pada dataset dilakukan berdasarkan tingkat kepentingan dan jumlah data yang hilang.
* Kolom seperti `director`, `cast`, dan `country` memiliki banyak missing value, sehingga tidak dihapus, melainkan diisi dengan nilai “Unknown” agar tidak kehilangan banyak data.

* Kolom `date_added` memiliki missing value yang sedikit dan penting untuk analisis waktu, sehingga baris yang mengandung missing value dihapus.
* Kolom `rating` diisi menggunakan nilai modus karena jumlah missing sedikit dan masih dapat diperkirakan.
* Kolom penting seperti `title`, `type`, dan `release_year` tidak boleh memiliki missing value, sehingga jika terdapat nilai kosong maka baris tersebut dihapus.

## Data Transformation
1. buatlah feature engineering dengan membuat kolom baru dengan nama **year_added** dari kolom **date_added**

````python
# Menambahkan tahun. bulan, hari
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
df['day_added'] = df['date_added'].dt.day

df.head()
````

2. Buat Kolom baru atau feature engineering dengan nama kolom **kategori_konten** dari kolom **release_year** sesuai dengan kategori umur konten dibawah ini
```
jika tahun kurang dari 2000 kategori OLD, selain itu jika tahun kurang dari 2015 MIDDLE selain itu NEW
```



````python
# Membuat function untuk menentukan kategori konten berdasarkan tahun rilis (release_year)
def kategori_konten(year):
    if year < 2000:
        return 'OLD'
    elif year < 2015:
        return 'MIDDLE'
    else:
        return 'NEW'

# Menambahkan kolom kategori_konten dari kolom relase_year
df['kategori_konten'] = df['release_year'].apply(kategori_konten)

# Cek hasil
df[['release_year', 'kategori_konten']].head()
````

3. Encoding Kolom **Rating / Type**

````python
# Encoding kolom rating/type
df_encoded = pd.get_dummies(df, columns=['rating', 'type'], prefix=['rating', 'type'])

# Cek hasilnya
df_encoded.head()
````
